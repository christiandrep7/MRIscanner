# MRIscanner: train on genuinely external data (BraTS + IXI)

The Kaggle "Brain Tumor MRI Dataset" this project trains on is, per its own
documented provenance, a combination of just 3 source collections from a narrow
set of scanners -- models trained on it generalize poorly to real-world scans
from elsewhere (works on Kaggle-style test scans, struggles on real online
photos). This notebook adds genuinely independent data to the 2 classes where a
free, no-registration source actually exists:

- **glioma**: [BraTS](https://www.med.upenn.edu/cbica/brats2020/data.html) -- 19 real
  institutions, T1-contrast-enhanced scans + tumor segmentation masks.
- **notumor**: [IXI](https://brain-development.org/ixi-dataset/) -- 3 real hospitals
  (Hammersmith/Guys/Institute of Psychiatry), healthy control scans.

meningioma/pituitary are **not** extended here: the only independent alternative
found (TCIA's Meningioma-SEG-CLASS) requires *you* to personally sign a
restricted-license data-use agreement with TCIA -- a real per-person legal step
that can't be done on your behalf from a notebook.

### BraTS (glioma): download, extract, free the raw volumes
Only t1ce + seg are ever used, but the Kaggle API downloads full datasets in one call, so this deletes the other 3 modalities (+ the leftover zip) right after extraction -- and then, once glioma slices are extracted to PNG, deletes the raw volumes entirely **before** IXI's download starts. Real bug hit running an earlier version of this notebook on Colab: BraTS + IXI's raw volumes both sitting on disk at once exhausted a 113GB disk (`OSError: [Errno 28] No space left on device` / Colab's own low-disk warning). Processing one dataset fully (download -> extract -> delete raw) before starting the next keeps peak usage to whichever single dataset is largest, not both combined.

```python
import shutil
from pathlib import Path
from kaggle.api.kaggle_api_extended import KaggleApi
from run_external_data_pipeline import _reclaim_disk_after_brats_download
from src.external_data import extract_brats_glioma_slices

api = KaggleApi()
api.authenticate()

brats_dir = Path("/content/external_raw/brats")
brats_dir.mkdir(parents=True, exist_ok=True)

print("Downloading BraTS2020 (~7GB compressed)...")
api.dataset_download_files(
    "awsaf49/brats20-dataset-training-validation",
    path=str(brats_dir), unzip=True, quiet=False,
)
_reclaim_disk_after_brats_download(brats_dir)

print("Extracting glioma slices from BraTS...")
n_glioma = extract_brats_glioma_slices(
    brats_dir, Path("/content/external_raw_slices/glioma"), slices_per_patient=6
)
print(f"Extracted {n_glioma} new glioma slices from BraTS")
assert n_glioma > 100, (
    f"Only {n_glioma} glioma slices extracted -- something's off with the BraTS "
    "download/layout (check the unzip above actually produced BraTS20_Training_* folders)."
)

shutil.rmtree(brats_dir, ignore_errors=True)
print("Freed BraTS raw volumes (already sliced to PNG).")
```

### IXI (notumor): download, extract, free the raw volumes
Same pattern as BraTS -- fully processed and its raw volumes freed before anything else needs the disk. Both reoriented to canonical (RAS+) before slicing -- verified during development that skipping this step can silently produce a sagittal slice instead of axial.

```python
from run_external_data_pipeline import _reclaim_disk_after_ixi_download
from src.external_data import extract_ixi_notumor_slices

ixi_dir = Path("/content/external_raw/ixi")
ixi_dir.mkdir(parents=True, exist_ok=True)

print("Downloading IXI (~4GB)...")
api.dataset_download_files(
    "wailrami/ixi-healthy-brain-mri-t1-t2",
    path=str(ixi_dir), unzip=True, quiet=False,
)
_reclaim_disk_after_ixi_download(ixi_dir)

print("Extracting notumor slices from IXI...")
n_notumor = extract_ixi_notumor_slices(
    ixi_dir, Path("/content/external_raw_slices/notumor"), slices_per_subject=4
)
print(f"Extracted {n_notumor} new notumor slices from IXI")
assert n_notumor > 100, (
    f"Only {n_notumor} notumor slices extracted -- something's off with the IXI "
    "download/layout (check the unzip above actually produced .nii files)."
)

shutil.rmtree(ixi_dir, ignore_errors=True)
print("Freed IXI raw volumes (already sliced to PNG).")
```

### Dedupe against the existing dataset
BraTS/IXI are independent sources, so real overlap is very unlikely -- but this reuses the project's own MD5 + perceptual-hash dedup (built for exactly this kind of check) as cheap insurance against test-set leakage before anything gets merged into data/Training.

```python
from src.secret_dedupe_eval import build_reference_index, dedupe_secret

reference_md5, reference_phash = build_reference_index([Path("data/Training"), Path("data/Testing")])
stats = dedupe_secret(
    secret_root=Path("/content/external_raw_slices"),
    out_root=Path("/content/external_deduped"),
    md5_set=reference_md5,
    ph_list=reference_phash,
    max_hamming=5,
)
print(stats)
```

### Merge by PATIENT: most patients into Training, whole other patients into a fresh ExternalTesting set
Each patient contributes several slices. **All slices of a patient go to the same side** -- an earlier version of this notebook shuffled individual slices, which put neighbouring slices of the same brain in both Training and ExternalTesting (100% of the old external test slices had a same-patient slice in training), inflating the score. `data/ExternalTesting` now contains only patients the model never saw a single slice of.


```python
from src.external_data import find_patient_overlap, split_by_patient

EXTERNAL_TEST_FRACTION = 0.2

for class_name in ["glioma", "notumor"]:
    src_dir = Path("/content/external_deduped") / class_name
    if not src_dir.exists():
        print(f"No deduped images for {class_name}, skipping")
        continue
    n_train, n_test = split_by_patient(
        src_dir,
        Path("data/Training") / class_name,
        Path("data/ExternalTesting") / class_name,
        test_fraction=EXTERNAL_TEST_FRACTION,
        seed=42,
    )
    print(f"{class_name}: {n_train} slices -> data/Training, {n_test} slices -> data/ExternalTesting")

overlap = find_patient_overlap(Path("data/Training"), Path("data/ExternalTesting"))
assert not overlap, f"Patient leakage: {len(overlap)} patients on both sides, e.g. {sorted(overlap)[:5]}"
print("verified: 0 patients appear in both Training and ExternalTesting")
```

### Remove train/test twins (the Kaggle set has no patient ids)
Measured on the original data: ~27% of the 1,600 'held-out' Kaggle Testing images had a near-identical copy (same scan, re-saved as a different file) in Training. Those twins are moved out of **Training** into `data/_quarantined_duplicates_of_test` (never read by training). The test sets are left untouched, so every image tested on is one the model has not seen a copy of.


```python
from src.leakage import quarantine_train_duplicates_of_test, find_test_images_with_train_duplicates

quarantine = Path("data/_quarantined_duplicates_of_test")
for test_name in ("Testing", "ExternalTesting"):
    moved = quarantine_train_duplicates_of_test(Path("data/Training"), Path("data") / test_name, quarantine / test_name)
    print(f"{test_name}: moved {moved} near-duplicate training images to quarantine")

for test_name in ("Testing", "ExternalTesting"):
    left = find_test_images_with_train_duplicates(Path("data/Training"), Path("data") / test_name)
    assert not left, f"{test_name}: {len(left)} test images still have a training twin"
print("verified: no test image has a near-duplicate in Training")
```

```python
# Sanity check: final class counts across both training and both test sets.
for split in ["Training", "Testing", "ExternalTesting"]:
    split_dir = Path("data") / split
    if not split_dir.exists():
        continue
    print(f"\n{split}:")
    for class_dir in sorted(p for p in split_dir.iterdir() if p.is_dir()):
        count = sum(1 for _ in class_dir.glob("*"))
        print(f"  {class_dir.name}: {count}")
```

### Benchmark against data/ExternalTesting -- brand-new patients
These are BraTS/IXI **patients** the models never saw any slice of (the split is by patient, verified above), so this measures recognizing strangers, not recognizing familiar brains. Same hospitals/scanners as training, though -- this is *new patients*, not *new institutions*.


```python
from src.config import DataConfig
from src.benchmark import default_checkpoint_specs, benchmark_models, save_benchmark_report

external_cfg = DataConfig(test_dir_name="ExternalTesting")
specs = default_checkpoint_specs(Path("checkpoints"))
external_results = benchmark_models(specs, external_cfg, output_dir=Path("outputs/benchmark_external"))
save_benchmark_report(external_results, Path("outputs/benchmark_external"))
for r in external_results:
    if r.available:
        print(f"{r.architecture}: accuracy={r.accuracy:.4f} macro_f1={r.macro_f1:.4f}")
    else:
        print(f"{r.architecture}: unavailable ({r.error})")
```

### Glioma numbers side by side: original Testing vs the new external-only ExternalTesting

```python
import json

for label, bench_dir in [("data/Testing", "outputs/benchmark"), ("data/ExternalTesting", "outputs/benchmark_external")]:
    print(f"--- {label} ---")
    for arch in ["resnet50", "efficientnet_b0", "vgg16"]:
        report_path = Path(bench_dir) / arch / "classification_report.json"
        if not report_path.exists():
            continue
        report = json.load(open(report_path))
        g = report.get("glioma")
        if g:
            print(
                f"  {arch:16s} glioma recall={g['recall']:.4f}",
                f"precision={g['precision']:.4f}",
                f"overall_acc={report['accuracy']:.4f}",
            )
```
